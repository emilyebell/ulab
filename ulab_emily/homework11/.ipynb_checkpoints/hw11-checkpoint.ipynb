## 1 Functions
**Orbital Periods and Kepler's Third Law** \
Kepler's Third Law states that the square of the orbital period $T$ (in years) of a planet is proportional to the cube of the semi-major axis $a$ (in astronomical units, AU):

$$
T^2 = a^3
$$

Write a **function** that calculates the orbital period $T$ given the semi-major axis $a$. (So, you can assume that $T = \sqrt{a^3}$.)

```python
# Your code here (WITHOUT NUMPY! or any other modules/libraries/packages)

def orbital_period(a):
    # calculate oribital period given semi-major axis (a)
    return (a ** 3) ** 0.5  
```

Call your function with the following values: 
* `planet_a` $ = 0.5\,\text{AU}$
* `planet_b` $ = 1.0\,\text{AU}$
* `planet_c` $ = 1.5\,\text{AU}$

```python
# Your code here
planet_a = 0.5  #AU
planet_b = 1.0 #AU
planet_c = 1.5 #AU

T_a = orbital_period(planet_a)
T_b = orbital_period(planet_b)
T_c = orbital_period(planet_c)

print(f"Orbital period of Planet A: {T_a:.2f} years")
print(f"Orbital period of Planet B: {T_b:.2f} years")
print(f"Orbital period of Planet C: {T_c:.2f} years")
```

Rewrite your **function** from above with a `numpy` function. 

```python
# Your code here
def orbital_period_numpy(a):
    # calculate oribital period given semi-major axis (a)
    return np.sqrt(np.power(a, 3))  
```

Call both of your functions in the cell below with ($a$ = 5). Time how long it takes each of your functions to run. **DO NOT REWRITE THEM!** 

```python
# Your code here
a_value = 5

start_time_np = time.time()
result_np = orbital_period_numpy(a_value)
end_time_np = time.time()
time_np = end_time_np - start_time_np

start_time = time.time()
result = orbital_period(a_value)
end_time = time.time()
time_og = end_time - start_time

print(f"Orbital period of a = 5 (original function): {result:.2f} years")
print(f"Original Function Time: {time_og}")
print(f"Orbital period of a = 5 (numpy function): {result:.2f} years")
print(f"Numpy Function Time: {time_np}")
```

## 2 Indexing + Conditionals

You are provided with a list of exoplanet radii (in Earth radii) and their equilibrium temperatures (in Kelvin):

`radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]` \
`temperatures = [300, 400, 350, 280, 500, 450]`

Write a loop to filter and print the radii of planets with temperatures below 350 K.

```python
# Your code here

radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]
temperatures = [300, 400, 350, 280, 500, 450]

for i in range(len(temperatures)):
    if temperatures[i] < 350:
        print(f"planet {i+1}: radius = {radii[i]} Earth radii, Temperature = {temperatures[i]} K")
```

Use list slicing to extract and print the radii of the first three planets.

```python
# Your code here

first_3 = radii[:3]
print(f"First 3 radii:{first_3}")
```

Turn the radii and temperatures data into a dictionary. 

```python
# Your code here

planets = dict(zip(radii, temperatures)) #assign radii and temp to planets
print("Exoplanet data:", planets)
```

Give each of the planets in your dictionary a name. 

```python
# Your code here

planet_names = ['Planet 1', 'Planet 2', 'Planet 3', 'Planet 4', 'Planet 5', 'Planet 6']

planets = {planet_names[i]: (radii[i], temperatures[i]) for i in range(len(planet_names))}
print('Planet data w names:', planets)
```

Create a new dictionary of planets with radii between 1.0 and 2.0 Earth radii (inclusive).

```python
# Your code here

planets_filtered = {name: data for name, data in planets.items() if 1.0 <= data[0] <= 2.0}
print("Filtered planets (1.0-2.0 Earth radii):", planets_filtered)
```

## 4 Plotting
Create a scatter plot of the following semi-major axes (x-axis) vs. orbital periods (y-axis). Label the axes, add a legend and add a title: "Orbital Periods of Exoplanets."

`semi_major_axes = [0.5, 1.0, 2.0, 1.5, 1.2, 0.8]`

```python
# Your code here
import matplotlib.pyplot as plt

semi_major_axes = [0.5, 1.0, 2.0, 1.5, 1.2, 0.8]

orbital_periods = orbital_period_numpy(semi_major_axes)

plt.scatter(semi_major_axes, orbital_periods, label = "orbital periods")
plt.xlabel('semi-major axes (AU)')
plt.ylabel('orbital periods (years)')
plt.title('orbital periods of exoplanets')
plt.legend()
plt.show()
```

Create a bar plot of planet radii vs. their equilibrium temperatures. Label the axes, add a legend and add a title: "Equilibrium Temperatures vs. Radii."

```python
# Your code here
radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]
temperatures = [300, 400, 350, 280, 500, 450]
plt.bar(radii, temperatures, label='eq. temperatures')
plt.xlabel('planet radii (Earth Radii)')
plt.ylabel('eq. temperature (K)')
plt.title("Eq. Temperatures vs Radii")
plt.legend()
plt.show()
```

Put the scatter plot and bar plot on the same figure (horizonatally). 

```python
# Your code here
fig, (ax1, ax2) = plt.subplots(1,2, figsize=(12,6))
#scatter
ax1.scatter(semi_major_axes, orbital_periods, label='orbital periods')
ax1.set_xlabel('semi-major axes (AU)')
ax2.set_ylabel('orbital period (years)')
ax1.set_title('orbital periods of exoplants')
ax1.legend()

#barplot
ax2.bar(radii, temperatures, label='eq temperatuers')
ax2.set_ylabel('eq temperature (K)')
ax2.set_title('Eq. Temperature vs Radii')
ax2.legend()

plt.tight_layout()
plt.show()
```
